```python
tasks = []

def add_task(task):
    """
    Add a task to the tasks list

    >>> tasks_test = []
    >>> tasks_test.append("Buy milk")
    >>> len(tasks_test)
    1
    """
    tasks.append(task)
    print(f"Added: {task}")

def remove_task(task_number):
    """
    Remove a task at the given task number (1-based, as shown to user)

    >>> tasks_test = ["Task 1", "Task 2", "Task 3"]
    >>> task_num = 2  # User wants to remove task #2
    >>> index = task_num - 1  # Convert to 0-based index
    >>> if 0 <= index < len(tasks_test):
    ...     removed = tasks_test[index]
    ...     del tasks_test[index]
    ...     print(f"Removed: {removed}")
    Removed: Task 2
    >>> len(tasks_test)
    2
    """
    # Fixed: Changed 'tas' to 'tasks' on line 8
    # Fixed: Convert task_number (1-based) to index (0-based)
    index = task_number - 1

    if index >= 0 and index < len(tasks):
        removed = tasks[index]
        del tasks[index]
        print(f"Removed: {removed}")
    else:
        print("Invalid task number!")

def list_tasks():
    """
    Display all tasks in the list with 1-based numbering

    >>> tasks_test = ["Buy milk", "Walk dog"]
    >>> for i, t in enumerate(tasks_test):
    ...     print(f"{i+1}: {t}")
    1: Buy milk
    2: Walk dog
    """
    if not tasks:
        print("No tasks available.")
    else:
        # Fixed: Changed 'task' to 'tasks' on line 19
        for i, t in enumerate(tasks):
            print(f"{i+1}: {t}")  # Display with 1-based numbering

def main():
    """Main program to demonstrate task manager functionality"""
    # Add some sample tasks
    add_task("Buy milk")
    add_task("Pay bills")
    add_task("Walk dog")

    # List all tasks (shows 1: Buy milk, 2: Pay bills, 3: Walk dog)
    print("\nCurrent tasks:")
    list_tasks()

    # Remove task #2 (Pay bills) - now correctly removes the second displayed task
    print("\nRemoving task #2 (Pay bills):")
    remove_task(2)

    # Fixed: Changed 'list_task()' to 'list_tasks()' on line 29
    print("\nRemaining tasks:")
    list_tasks()

if __name__ == "__main__":
    import doctest
    doctest.testmod()
    main()
```
